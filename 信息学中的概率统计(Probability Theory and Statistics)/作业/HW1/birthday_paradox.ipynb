```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def arrange_num(m, n): # calculate m(m-1)(m-2)...(m-n+1)
    result = 1
    for i in range(n):
        result *= m - i
    return result

def power(m, n): # calculate m^n
    result = 1
    for _ in range(n):
        result *= m
    return result

def prob_collision(m, n): # calculate the probability of collision
    return 1 - arrange_num(m, n) / power(m, n)
```

```python
n = np.arange(2, 51)
ans_m = []
```

```python
for n in range(2, 51):
    for m in range(1, 1000000):
        if prob_collision(m, n) <= 0.01:
            print("n = %d, m = %d" % (n, m))
            ans_m.append(m)
            break
```

```python
ans_m
```

```python
# visualize the result
plt.figure(figsize=(10, 6))
plt.plot(np.arange(2, 51), ans_m)
plt.plot(np.arange(2, 51), ans_m, 'go')
plt.plot(np.arange(2, 51), 49*np.arange(2, 51)**2, 'b.')
plt.xlabel('n')
plt.ylabel('m')
plt.title('m that makes the probability of collision <= 0.01')
plt.savefig('collision.pdf')
```
