```python
def factorial(n):
    # if n == 1:
    #     return 1
    # return n * factorial(n-1)
    res = 1
    for i in range(1, n+1):
        res *= i
    return res

def combination(n, m):
    return factorial(n) / (factorial(m) * factorial(n-m))
```

```python
res = 0
for i in range(2):
    print(combination(1000, i) * (0.005 ** i) * (0.995 ** (1000 - i)))
    res += combination(1000, i) * (0.005 ** i) * (0.995 ** (1000 - i))
res
```
